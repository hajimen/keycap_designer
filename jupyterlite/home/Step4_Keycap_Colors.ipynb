```python
# Boilerplate for JupyterLite 
import install_keycap_designer
from keycap_designer.jupyter import *
```

```python
# Preparation
s_center_red = Style(
    size=2.5,
    x_loc=0.,
    y_loc=0.,
    font='OpenSans-VariableFont_wdth,wght.ttf',
    h_o=Center,
    align=Center,
    v_o=Center,
    color= sRGBColor('#FF0000'),
    side=TopSide
)
legend_center_red = Legend({s_center_red: 'Red'})
```

```python
# The big advantage of Junana proifile is keycap color.
# Dye sublimation printing on black keycaps!
# You can specify it like this:
Junana @ Specifier('1u') @ legend_center_red @ BackgroundColor(sRGBColor(6, 6, 6))
# Why `sRGBColor(6, 6, 6)`, not `sRGBColor(0, 0, 0)`? There is a long long story behind that...
```

```python
# You can specify color just for top side:
Junana @ Specifier('1u') @ legend_center_red @ SideColor({TopSide: sRGBColor(6, 6, 6)})
```

```python
# You can also do like this:
Junana @ Specifier('1u') @ legend_center_red @ BackgroundColor(sRGBColor(0, 255, 0)) @ SideColor({TopSide: sRGBColor(6, 6, 6)})
```

```python
# You can fill the background by an image tiling:
Junana @ Specifier('1u') @ legend_center_red @ Wallpaper('wallpaper_pattern.png')
```

```python
# By specifying a large image to `Wallpaper()`, you can put an image file into five sides:
Junana @ Specifier('1u') @ legend_center_red @ Wallpaper('mandrill_large.jpg')
```
