```python
# Boilerplate for JupyterLite 
import install_keycap_designer
from keycap_designer.jupyter import *
```

```python
# Let's load an image file. You can find the 'mandrill.jpg' in the left pane.
top_image = TopImage('mandrill.jpg')
```

```python
# You can ckeck the loaded image by inline view:
top_image
```

```python
# Specify keycap profile and variation. They are mandatory for a manuscript.
# What is 'manuscript'? Printing-ready design data.
# Variation '1u' is the most common one.
m_mandatory = Junana @ Specifier('1u')
# What is the `@`? It is an operator to make a manuscript.
```

```python
# You can check the manuscript by inline view:
m_mandatory
```

```python
# Look at the inline view above.
# The artwork image shows a blank keycap because `m_mantadory` has nothing to print.
```

```python
# Let's combine the mandrill image with `m_mandatory`:
m_mandrill = m_mandatory @ top_image
m_mandrill
```

```python
# Is the artwork image too small? Higher DPI is available.
change_default_dpi(720)  # Default is 192.
m_mandrill
```

```python
# To order custom printed keycaps, please generate a PDF and send it to DecentKeyboards:
generate_pdf(m_mandrill, 'mandrill.pdf')
# Now you can find the 'mandrill.pdf' in the left pane.
```

```python
# If you need multiple keycaps with the same design, use `Repeat`:
m_mandrill_5pcs = m_mandrill @ Repeat(5)
generate_pdf(m_mandrill_5pcs, 'mandrill_5pcs.pdf')
```
