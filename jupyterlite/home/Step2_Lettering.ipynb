```python
# Boilerplate for JupyterLite 
import install_keycap_designer
from keycap_designer.jupyter import *
```

```python
# This is the same manuscript with the step 1:
m_mandrill = Junana @ Specifier('1u') @ TopImage('mandrill.jpg')
```

```python
# Style specifies font, font size, location, color, etc.
s_center_red = Style(
    size=2.5, # font size. The unit is mm.

    # [xy]_loc are the distance from the origin. The origin is specified by [hv]_o and align.
    # The unit is mm.
    x_loc=0.,
    y_loc=0.,

    font='OpenSans-VariableFont_wdth,wght.ttf',
    # You can use variable fonts. The 'NotoSansMono-VariableFont_wdth,wght.ttf' is also a variable font.

    # Layout to center of the printable area.
    # h denotes 'horizontal', v denotes 'vertical', and o denotes 'origin'.
    h_o=Center,
    align=Center,
    v_o=Center,

    # Legend's color
    color= sRGBColor('#FF0000'),

    # Print on top
    side=TopSide
)
```

```python
# Legend specifies style and string
legend_center_red = Legend({s_center_red: 'Red'})
m_mandrill @ legend_center_red
```

```python
# You can make a Style object from existing object
s_left_top_blue = s_center_red.mod(h_o=Left, align=Left, v_o=Top, color=sRGBColor('#0000FF'))
```

```python
# The `s_left_blue` has different location and color, and everything else is the same
legend_left_top_blue = Legend({s_left_top_blue: 'Blue'})
m_mandrill @ legend_left_top_blue
```

```python
# You can combine legends by `@` operator
m_mandrill @ legend_left_top_blue @ legend_center_red
```

```python
# Or like this
m_mandrill @ Legend({s_center_red: 'Red', s_left_top_blue: 'Blue'})
```

```python
# "Blue" is cropped. Let's fix it by `mod()`.
s_left_top_blue_uncropped = s_left_top_blue.mod(x_loc=1.)
m_mandrill @ Legend({s_left_top_blue_uncropped: 'Blue'})
```
