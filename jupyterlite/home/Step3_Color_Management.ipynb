```python
# In step 1 and 2, you might find the mandrill's colors are pale in manuscripts.
# It comes from color management.
# Unlike LCDs, keycap printing cannot produce vivid colours.
# We should manage to do with the limitation. Color management does it.
# Colors in manuscripts are a simulation of actual printing colors.

# We can choose from several ways of color management.
# First, the default way, `Perceptual`:
m_mandrill = Junana @ Specifier('1u') @ TopImage('mandrill.jpg')
m_mandrill
```

```python
# `Relative`:
m_mandrill @ Relative
```

```python
# `RelativeNoBpc`:
m_mandrill @ RelativeNoBpc
```
