```python
# Boilerplate for JupyterLite 
import install_keycap_designer
from keycap_designer.jupyter import *
```

```python
# Of course you can print on front
Junana @ Specifier('1u') @ FrontImage('mandrill.jpg')
```

```python
# To print legends, specify the side in Style
s_center_front = Style(
    size=4.,
    x_loc=0.,
    y_loc=0.,
    font='OpenSans-VariableFont_wdth,wght.ttf',
    h_o=Center,
    align=Center,
    v_o=Center,
    color= sRGBColor('#000000'),

    # here
    side=FrontSide
)
Junana @ Specifier('1u') @ Legend({s_center_front: 'm'})
```

```python
# Of course you can print on both
s_center_top = s_center_front.mod(side=TopSide)
Junana @ Specifier('1u') @ Legend({s_center_top: 'M', s_center_front: 'm'})
```
